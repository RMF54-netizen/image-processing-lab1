# NumPy Exercises

Now that we've learned about NumPy let's test your knowledge. We'll start off with a few simple tasks, and then you'll be asked some more complicated questions.

#### Import NumPy as np

```python
import numpy as np
```

#### Create an array of 10 zeros

```python
zeros_arr = np.zeros(10)
zeros_arr
```

#### Create an array of 10 ones

```python
ones_arr = np.ones(10)
ones_arr
```

#### Create an array of 10 fives

```python
fives_arr = np.ones(10) * 5
fives_arr
```

#### Create an array of the integers from 10 to 50

```python
int_arr = np.arange(10, 51)
int_arr
```

#### Create an array of all the even integers from 10 to 50

```python
even_arr = np.arange(10, 51, 2)
even_arr
```

#### Create a 3x3 matrix with values ranging from 0 to 8

```python
matrix_3x3 = np.arange(0, 9).reshape(3, 3)
matrix_3x3
```

#### Create a 3x3 identity matrix

```python
identity_3x3 = np.eye(3)
identity_3x3
```

#### Use NumPy to generate a random number between 0 and 1

```python
random_num = np.random.rand(1)
random_num
```

#### Use NumPy to generate an array of 25 random numbers sampled from a standard normal distribution

```python
random_25 = np.random.randn(25)
random_25
```

#### Create an array of 20 linearly spaced points between 0 and 1:

```python
linspace_arr = np.linspace(0, 1, 20)
linspace_arr
```

### Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

```python
import numpy as np

arr_2d = np.array([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
], np.uint8)

arr_2d
```

Retrieve the value 25 from the array using indexing.

```python
arr_2d[1, 1]
```

Extract the second row of the array.

```python
arr_2d[1]
```

Extract the third column of the array.

```python
arr_2d[:, 2]
```

Extract the bottom-right 2×2 sub-array from arr_2d.

```python
arr_2d[1:3, 1:3]
```
