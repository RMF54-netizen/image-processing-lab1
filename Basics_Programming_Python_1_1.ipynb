### Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

````python
import numpy as np

arr_2d = np.array([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
], np.uint8)

arr_2d
````

Retrieve the value 25 from the array using indexing.

````python
arr_2d[1, 1]
````

Extract the second row of the array.

````python
arr_2d[1]
````

Extract the third column of the array.

````python
arr_2d[:, 2]
````

Extract the bottom-right 2×2 sub-array from arr_2d.

````python
arr_2d[1:, 1:]
````

````python
import os
os.makedirs('images', exist_ok=True)

if not os.path.exists('images/cameraman.png') or not os.path.exists('images/lena_gray_256.png'):
    import cv2
    import numpy as np
    import urllib.request
    from PIL import Image
    from skimage import data

    cam = cv2.resize(data.camera(), (256, 256))
    cv2.imwrite('images/cameraman.png', cam)

    url = 'https://raw.githubusercontent.com/mikolalysenko/lena/master/lena.png'
    urllib.request.urlretrieve(url, 'lena_temp.png')
    lena = Image.open('lena_temp.png').convert('L').resize((256, 256))
    lena.save('images/lena_gray_256.png')
    os.remove('lena_temp.png')

    print('images ready')
````
